```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("bigquery://olist-data-pipeline-507001")


```

```python
query = """
SELECT
    DATE(o.order_purchase_timestamp) AS sales_date,
    SUM(oi.price) AS total_sales
FROM `olist-data-pipeline-507001.olist_mart.fact_orders` o
JOIN `olist-data-pipeline-507001.olist_mart.fact_order_items` oi
    ON o.order_key = oi.order_key
WHERE o.order_status = 'DELIVERED'
GROUP BY sales_date
ORDER BY sales_date
"""

daily_sales = pd.read_sql(query, engine)

daily_sales.head()
```

```python
query = """
SELECT
    MIN(DATE(order_purchase_timestamp)) AS first_date,
    MAX(DATE(order_purchase_timestamp)) AS last_date
FROM `olist-data-pipeline-507001.olist_mart.fact_orders`
"""

date_range = pd.read_sql(query, engine)

date_range
```

```python
query = """
SELECT *
FROM `olist-data-pipeline-507001.olist_staging.stg_holiday_calendar`
"""

holidays = pd.read_sql(query, engine)
holidays
```

```python
holidays.head()
```

```python
holidays["hol_date"].min(), holidays["hol_date"].max()

daily_sales["sales_date"].dtype, holidays["hol_date"].dtype

daily_sales["sales_date"] = pd.to_datetime(daily_sales["sales_date"])
holidays["hol_date"] = pd.to_datetime(holidays["hol_date"])
```

```python
sales_holidays = daily_sales.merge(
    holidays,
    left_on="sales_date",
    right_on="hol_date",
    how="left"
)

sales_holidays.head()
```

```python
sales_holidays["is_holiday"] = sales_holidays["hol_name"].notna()

sales_holidays.head()
```

```python
sales_holidays[sales_holidays["is_holiday"] == True].head(10)
```

```python
sales_holidays.groupby("is_holiday")["total_sales"].mean()
```

```python
sales_holidays["is_holiday"].value_counts()
```

```python
sales_holidays[sales_holidays["is_holiday"] == True][
    ["sales_date", "hol_name", "total_sales"]
].sort_values("total_sales")
```

```python
holiday_sales = (
    sales_holidays[sales_holidays["is_holiday"] == True]
    .groupby("hol_name")["total_sales"]
    .agg(["mean", "min", "max", "count"])
    .sort_values("mean")
)

holiday_sales
```

```python
#Are holiday sales lower because it is a holiday, or because holidays happen to fall on days when sales are normally lower?
daily_sales["day_of_week"] = daily_sales["sales_date"].dt.day_name()

weekday_sales = (
    daily_sales
    .groupby("day_of_week")["total_sales"]
    .mean()
    .sort_values()
)

weekday_sales
```

```python
sales_holidays["day_of_week"] = sales_holidays["sales_date"].dt.day_name()
```

```python
holiday_comparison = sales_holidays[sales_holidays["is_holiday"] == True].copy()

holiday_comparison["normal_weekday_sales"] = (
    holiday_comparison["day_of_week"].map(weekday_sales)
)

holiday_comparison["sales_difference"] = (
    holiday_comparison["total_sales"]
    - holiday_comparison["normal_weekday_sales"]
)

holiday_comparison[
    ["sales_date", "hol_name", "day_of_week",
     "total_sales", "normal_weekday_sales", "sales_difference"]
].sort_values("sales_difference")
```
